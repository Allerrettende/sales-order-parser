```python
import re

# 日期格式: 3.10.2026 或 03/10/2026 或 3-10-2026 等
date_pattern = r'\d{1,2}[./-]\d{1,2}[./-]\d{4}'

# 测试
dates = [
    "3.10.2026",      # ✅
    "03/10/2026",     # ✅
    "3-10-2026",      # ✅
    "12.05.2024",     # ✅
    "1/1/2025",       # ✅
    "19-12/2023",     # ✅
]

for date in dates:
    match = re.search(date_pattern, date)
    print(f"{date}: {match.group(0) if match else '不匹配'}")
```
