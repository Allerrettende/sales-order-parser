```python
import pandas as pd
import numpy as np

# 1. pd.NA 也是 False
# print(bool(pd.NA))        # False ✅

# 2. np.nan 也是 False
print(bool(np.nan))       # False ✅

# 3. 但无法直接用于 if 判断（会报错）
try:
    if pd.Series([1, 2, 3]) > 0:
        print("OK")
except ValueError as e:
    print(f"错误: {e}")   # 报错：布尔值不确定

# 4. 正确做法：使用 .any() 或 .all()
s = pd.Series([1, 2, 3])
if s.gt(0).all():         # 所有值都 > 0
    print("全部大于0")    # 输出: 全部大于0

# 5. 判断单个 NaN
result = np.nan
if pd.isna(result):       # ✅ 正确
    print("是空值")
```

```python
import re

# 你的正则表达式

pattern = r"(^\d{5,7})\s*(\w*)?"
test_string = '201815 '

# 执行匹配
match = re.search(pattern, test_string)

if match:
    print(f"完整匹配: {match.group(0)}")  # 201815*
    print(f"组1: {match.group(1)}")       # 201815
    print(f"组2: {match.group(2)}")       # '' (空字符串)
    print(f"组2类型: {type(match.group(2))}")  # <class 'str'>
    
    # 判断
    if match.group(2) is None:
        print("是 None")
    else:
        print(f"是 {repr(match.group(2))}")  # 是 ''
        print(f"长度: {len(match.group(2))}")  # 0
```

```python
import re

def handle_regex_result(match):
    """正确处理正则匹配结果"""
    
    if match is None:
        # 没有匹配
        print("❌ 没有匹配")
        return None
    
    # 有匹配
    group1 = match.group(1)
    
    if group1 is None:
        # 捕获组没有匹配
        print("⚠️ 捕获组没有匹配")
        return None
    
    if group1 == '':
        # 捕获到空字符串
        print("⚠️ 捕获到空字符串")
        return None
    
    # 有实际内容
    print(f"✅ 捕获: {group1}")
    return group1

# 测试
patterns = [
    (r'(\w*)', 'abc'),       # 捕获: 'abc'
    (r'(\w*)', ''),          # 捕获: ''
    (r'(\w+)', ''),          # 没有匹配 (re.search 返回 None)
]

for pattern, text in patterns:
    match = re.search(pattern, text)
    handle_regex_result(match)
```

```python
item_totals = (
    df.groupby("sales_order_no")["amount"]
      .sum()
)

item_discount_totals= (
    df.groupby("sales_order_no")["discount"]
      .sum()
)
subtotals = (
    df.groupby("sales_order_no")["subtotal_amount"]
      .first()
)

check = pd.DataFrame({
    "item_total": item_totals,
    "item_discount_total":item_discount_totals,
    "subtotal": subtotals
})

check["difference"] = (
    check["subtotal"] - (check["item_total"]+check["item_discount_total"])
)

check
```
