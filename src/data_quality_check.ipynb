```python
item_totals = (
    df.groupby("sales_order_no")["amount"]
      .sum()
)

item_discount_totals= (
    df.groupby("sales_order_no")["discount"]
      .sum()
)
subtotals = (
    df.groupby("sales_order_no")["subtotal_amount"]
      .first()
)

check = pd.DataFrame({
    "item_total": item_totals,
    "item_discount_total":item_discount_totals,
    "subtotal": subtotals
})

check["difference"] = (
    check["subtotal"] - (check["item_total"]+check["item_discount_total"])
)

check
```

```python
check[check["difference"].abs() > 0.01]
```
