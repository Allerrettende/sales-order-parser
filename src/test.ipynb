```python
import pandas as pd

# df = pd.read_excel(r"D:\OC\sales order list 20260814.xlsx")
df = pd.read_excel(r"D:\OC\sales order list.xlsx")
df.shape
```

```python
df["sales_order_no"].nunique()
```

```python
df["sales_order_no"].nunique()
df.groupby("sales_order_no").size()
df.isna().sum()
```

```python
item_totals = (
    df.groupby("sales_order_no")["amount"]
      .sum()
)

item_discount_totals= (
    df.groupby("sales_order_no")["discount"]
      .sum()
)
subtotals = (
    df.groupby("sales_order_no")["subtotal_amount"]
      .first()
)

check = pd.DataFrame({
    "item_total": item_totals,
    "item_discount_total":item_discount_totals,
    "subtotal": subtotals
})

check["difference"] = (
    check["subtotal"] - (check["item_total"]+check["item_discount_total"])
)

check
```

```python
check[check["difference"].abs() > 0.01]
```
